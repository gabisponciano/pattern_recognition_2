# Q1 — Matriz de covariância global

## Dados

```python
import numpy as np
import pandas as pd

df = pd.read_csv("wall+following+robot+navigation+data/sensor_readings_24.data", header=None)
X = df.iloc[:, :24].to_numpy().T  # p x N (amostras nas colunas)
p, N = X.shape
print(p, N)
```

## Métodos 1 a 4

```python
from covariancia import mcovar1, mcovar2, mcovar3, mcovar4
```

## Referência: `np.cov` com divisão por N

```python
Cref = np.cov(X, bias=True)  # bias=True -> divide por N (padrão divide por N-1)
```

## Norma de E = Cmy − Cref

```python
metodos = {"Método 1": mcovar1, "Método 2": mcovar2, "Método 3": mcovar3, "Método 4": mcovar4}

tabela = pd.DataFrame({
    "||E||_F": {nome: np.linalg.norm(f(X) - Cref) for nome, f in metodos.items()},
    "||E||_2": {nome: np.linalg.norm(f(X) - Cref, 2) for nome, f in metodos.items()},
})
tabela.loc["np.cov padrão (N-1)"] = [np.linalg.norm(np.cov(X) - Cref),
                                     np.linalg.norm(np.cov(X) - Cref, 2)]
tabela
```
